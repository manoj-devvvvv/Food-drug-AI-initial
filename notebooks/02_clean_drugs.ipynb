# Brick 2 - clean the drugs

**Gate:** A funnel shows every removal step and every kept SMILES re-parses.

The narrative report for this brick is `reports/brick2_funnel.md`. The cells below load the saved artifacts; they do not retrain anything.

```python
import json
from pathlib import Path
from IPython.display import Image, display, Markdown
import pandas as pd

ROOT = Path.cwd().parent
REPORTS = ROOT / "reports"
FIGURES = ROOT / "figures"

def md(name):
    p = REPORTS / name
    display(Markdown(p.read_text(encoding="utf-8"))) if p.exists() else print("missing", name)

def js(name, keys=None, indent=2):
    p = REPORTS / name
    if not p.exists():
        print("missing", name); return None
    d = json.loads(p.read_text(encoding="utf-8"))
    print(json.dumps({k: d[k] for k in keys} if keys else d, indent=indent)[:6000])
    return d
```

```python
js('brick2_funnel.json')
```
