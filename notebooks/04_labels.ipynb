# Brick 4 - label the ground truth

**Gate:** Rules and lexicon are saved; the 100-sentence audit is recorded; the nine case pairs are held out.

The narrative report for this brick is `reports/brick4_labels.md`. The cells below load the saved artifacts; they do not retrain anything.

```python
import json
from pathlib import Path
from IPython.display import Image, display, Markdown
import pandas as pd

ROOT = Path.cwd().parent
REPORTS = ROOT / "reports"
FIGURES = ROOT / "figures"

def md(name):
    p = REPORTS / name
    display(Markdown(p.read_text(encoding="utf-8"))) if p.exists() else print("missing", name)

def js(name, keys=None, indent=2):
    p = REPORTS / name
    if not p.exists():
        print("missing", name); return None
    d = json.loads(p.read_text(encoding="utf-8"))
    print(json.dumps({k: d[k] for k in keys} if keys else d, indent=indent)[:6000])
    return d
```

```python
md('brick4_labels.md')
print(pd.read_csv(REPORTS / 'sentence_audit_100.csv').head(10).to_string())
```
